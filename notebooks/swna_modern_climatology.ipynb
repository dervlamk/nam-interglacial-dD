```python
# lat/lon subsetting bounds
lon_min, lon_max = -125, -85
lat_min, lat_max = 10, 42

# --- ETOPO05 topography --- #
filen=f'{extern}/obs.etopo5.zsurf.nc'
etopo_full=xr.open_dataset(f'{filen}').ROSE / 1000 # convert from m to km
etopo_full.attrs['units'] = 'km'
etopo_full=longitude_flip(etopo_full)
etopo_land=etopo_full.where(etopo_full>0, np.nan) # mask bathymetry
etopo_land = etopo_land.rename({'ETOPO05_Y':'lat','ETOPO05_X':'lon'})

# --- OIPC isotopes --- #
filen='OIPC_monthly_data.nc'
oipc=xr.open_dataset(f'{extern}/{filen}').isotopes[::-1,:,:].rename({'Lat':'lat','Lon':'lon'})
# OIPC grid registration correction 
# The file's coordinates place the field about 0.5 deg east and 0.25 deg south of where it
# belongs. Plotted as-is the data sits southeast of the coastline -- most visible as a white gap
# down the west side of Baja with colour spilling into the Gulf.
#
# Measured, not guessed: OIPC is land-only (NaN over ocean), so its own NaN pattern traces a
# coastline. Fitting that against the Natural Earth 50m polygons cartopy actually draws raises
# coastal-cell agreement over this domain from 0.82 to 0.96, and the optimum is dlon=-0.50,
# dlat=+0.25 -- exactly 6 and 3 cells of the 1/12 deg grid. Fitting the Gulf of Mexico, the US
# Pacific Northwest, northern South America and the western Mediterranean independently recovers
# the same shift to within the fit's resolution, so this is a global property of the file, not a
# regional or projection artefact.
#
# This is an EMPIRICAL registration, not a documented correction, and the root cause in the
# source file is unknown. If a corrected OIPC file turns up, delete this and use it instead.
# Applied before subsetting so lon_min/lat_min above mean what they say.
OIPC_DLON, OIPC_DLAT = -0.50, 0.25
oipc=oipc.assign_coords(lon=oipc.lon + OIPC_DLON, lat=oipc.lat + OIPC_DLAT)

oipc=oipc.sel(lon=slice(lon_min,lon_max), lat=slice(lat_min,lat_max))
oipc=oipc.transpose('month','lat','lon')

# --- IMERG precipitation rate --- #
"""
filen='/Users/dervlamk/OneDrive/data/obs/satellite/imerg/imerg_V07_19980101-20241231_monthly_gn.nc'
ds=xr.open_dataset(f'{filen}').precipitation * 24 # convert from mm/hr to mm/day
ds=ds.sel(lon=slice(lon_min,lon_max), lat=slice(lat_min,lat_max)).groupby('time.month').mean(dim='time') # calculate monthly climatologies
del ds.attrs["units"]
imerg.attrs['Units'] = 'mm/day'
"""
filen='/Users/dervlamk/OneDrive/data/obs/satellite/imerg/imerg.gn.timeseries.2001-2018.nc'
ds=xr.open_dataset(filen).precipitation  * 24 # convert from mm/hr to mm/day
ds=ds.sel(lon=slice(lon_min,lon_max), lat=slice(lat_min,lat_max)).groupby('time.month').mean(dim='time') # calculate monthly climatologies
del ds.attrs["units"]
ds.attrs['Units'] = 'mm/day'
imerg=ds.transpose('month','lat','lon')
```

```python
# calculate seasonal dD and prec values (assumes OIPC data is already flux-weighted)

# init dicitionaries
dD = {}
prec = {}
pprec = {}

seasons=np.array(['djf','jfm','mam','jja','jas','jjas','son','ann'])

# sum annual total precip
annual_total_p = imerg.sum(dim="month")

# get seasonal means of obs data
for season in seasons:
    months = get_season(season)
    dD[season] = oipc.isel(month=months).mean(dim='month') # isotopes
    prec[season] = imerg.isel(month=months).mean(dim='month') # precip
    
    if season != 'ann':
        pprec[season] = imerg.isel(month=months).sum(dim='month') / annual_total_p
    else:
        pprec[season] = imerg / annual_total_p
        
    pprec[season].attrs['Units'] = '%'
    pprec[season].attrs['newname'] = 'percent of annual total'
```

```python
# --- REGIONAL MEAN OIPC dDp AT THE CORE SITES --- #

# The boxes are defined once, in scripts/py_functions/domain_funcs.py -> core_site_boxes(), and
# their bounds are read back off the polygons here rather than retyped. The region that is
# averaged and the region drawn on the isoscape maps below are therefore the same object and
# cannot drift apart.
#
# Why a box and not the grid cell at the core site: both cores are marine, and OIPC is a
# *terrestrial* isoscape -- it is NaN over ocean. The value at a core's own coordinates is
# either missing or a coastal edge artefact. The box samples the adjacent continental
# precipitation that sources the leaf wax delivered to the site, which is the quantity the
# downcore dDp record is actually being compared against.
#
# The average is consequently over land cells only: .weighted().mean() skips NaN and
# renormalises the weights across the valid cells. n_land reports how many those are -- worth a
# look if the boxes are ever moved, since a box pushed offshore would still return a number.
#
# cos(lat) area weighting, matching latitude_weighted_mean() in data_funcs.py. Across 3.5 deg of
# latitude it moves the answer by <0.06 per mil, but it costs nothing to be right. The seasonal
# means come from dD[] above, i.e. plain unweighted month means, per the repo convention that
# OIPC is already flux-weighted (CLAUDE.md, "Scientific invariants").

site_boxes = core_site_boxes()

oipc_site_dD = {site: {} for site in site_boxes}
n_land = {}

for site, poly in site_boxes.items():
    w, s, e, n = poly.bounds          # shapely: (min_lon, min_lat, max_lon, max_lat)
    n_land[site] = int(oipc.isel(month=0).sel(lon=slice(w, e), lat=slice(s, n)).notnull().sum())
    for season in seasons:
        box = dD[season].sel(lon=slice(w, e), lat=slice(s, n))
        weights = np.cos(np.deg2rad(box.lat))
        oipc_site_dD[site][season] = float(box.weighted(weights).mean(('lat', 'lon')))

# Guaymas = DSDP-480/479, Mazatlan = NH22P. These are the names the modern-value markers in
# dDwax_timeslice-means_timeseries.ipynb expect, e.g. Guaymas['jfm'], Mazatlan['jjas'].
Guaymas, Mazatlan = oipc_site_dD['Guaymas'], oipc_site_dD['Mazatlan']

# report
print('OIPC dDp [per mil] -- cos(lat)-weighted mean over the land cells of each core-site box\n')
print(f"{'season':>8}" + ''.join(f'{site:>12}' for site in site_boxes))
for season in seasons:
    print(f'{season:>8}' + ''.join(f'{oipc_site_dD[site][season]:>12.2f}' for site in site_boxes))
print()
for site, poly in site_boxes.items():
    w, s, e, n = poly.bounds
    total = oipc.isel(month=0).sel(lon=slice(w, e), lat=slice(s, n)).size
    print(f'{site:>9}: lon {w} to {e}, lat {s} to {n} -- '
          f'{n_land[site]} land cells of {total} ({100 * n_land[site] / total:.0f}%)')
```

```python
# Seasonal OIPC isoscapes, JFM vs JAS, with the two core-site sampling regions.
lon, lat = oipc.lon, oipc.lat

text_kw1={'color':'k', 'weight':'bold', 'size':12, 'ha':'left', 'va':'bottom'}
map_bnds=[-120., -95., 15., 35.]
icmap=cmo.solar
ilevels=np.linspace(-90, -10, 41)
inorm=mpl.colors.BoundaryNorm(ilevels, icmap.N)

fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(11,5), layout='constrained',
                       subplot_kw={'projection': proj})

for i, seas in enumerate(['jfm','jas']):
    # plot isoscapes
    cf=ax[i].pcolormesh(lon, lat, dD[seas],  cmap=icmap, norm=inorm, transform=trans)
    # plot topography
    ax[i].contour(etopo_land.lon, etopo_land.lat, etopo_land,
              levels=np.linspace(.8, 4, 5), linewidths=.5, colors='k', transform=trans)
    # title each panel based on season key
    ax[i].text(map_bnds[0]+1.5, map_bnds[3]+0.5, seas, **text_kw1)
    # add scatter points for core site locations
    ax[i].scatter(x=clons, y=clats, c='k', alpha=1, edgecolor='k', s=150, transform=trans, zorder=100)
    # core-site sampling regions, drawn the same way as the NAM domain above
    ax[i].add_geometries(list(core_site_boxes().values()), crs=trans,
                         fc='none', ec='k', lw=1, linestyle='--', zorder=9)
    # map settings
    ax[i].set_extent(map_bnds, crs=trans)
    ax[i].coastlines()
    ax[i].add_feature(cfeature.BORDERS)
    gl=ax[i].gridlines(crs=trans, lw=0.5, colors='black', alpha=1.0, linestyle='--',
                       zorder=10, draw_labels=True)
    gl.top_labels=False; gl.right_labels=False
    if i==1: gl.left_labels=False

cbar_ax1 = fig.add_axes([1.01, .1, 0.025, 0.8])
cbar1 = fig.colorbar(cf, ticks=np.arange(-100,10,10), orientation='vertical', extend='both', cax=cbar_ax1)
cbar1.set_label(u'OIPC $\delta$D$_{p}$ [‰]', weight='normal', labelpad=15, rotation=270)
cbar1.ax.tick_params(labelsize=10)

# save output
#plt.savefig(f'{opath}/oipc_seasonal_isoscapes.png', dpi=1200, bbox_inches='tight')
```
